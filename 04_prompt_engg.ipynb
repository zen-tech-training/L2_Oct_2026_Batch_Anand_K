```python
from langchain_ollama import ChatOllama

model = ChatOllama(model="gpt-oss:120b-cloud")
print('Model is ready to use!')
```

```python
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="Which is the capital of {country}?",
    input_variables=["country"]
)
prompt = prompt_template.format(country="France")

response = model.invoke(prompt)
print(response)
```

```python
# Zero shot prompting
from langchain_core.prompts import PromptTemplate

prompt_template = PromptTemplate(
    template="Translate 'What is your name?' into {language}?",
    input_variables=["language"]
)
prompt = prompt_template.format(language="Hindi")

response = model.invoke(prompt)
print(response)
```

```python
# One shot prompting

from langchain_core.prompts import PromptTemplate

one_shot_prompt = (
    f"Example:\n"
    f"Question: Translate 'Hello, How are you?' into Hindi.\n"
    f"Answer: Aap kaise ho?\n\n"
    f"Now answer this:\n"
    f"Question: Translate 'What is your name?' into Hindi.\n"
    f"Answer: \n"
)

prompt_template = PromptTemplate(
    template=one_shot_prompt
)
prompt = prompt_template.format()

response = model.invoke(prompt)
print(response)
```

```python
# Few shot prompting
few_shot_prompt = """
Examples:
Message: My laptop hangs frequently
Category: Laptop

Message: Food quality in the office canteen is poor
Category: Food

Classify the following message into the category:
Message: Bus driver is a rash driver.
Category:
"""

prompt_template = PromptTemplate(
    template=few_shot_prompt
)
prompt = prompt_template.format()

response = model.invoke(prompt)
print(response)
```
