```python
import tiktoken

tiktoken = tiktoken.get_encoding("cl100k_base")

#prompt = "Which is the capital city of India?"
prompt = "What is a Generative AI"

tokens = tiktoken.encode(prompt)
print(f"Number of tokens: {len(tokens)}")

for token in tokens:
    print(f"{tiktoken.decode([token])} - {token}")
```
